```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,        # in characters (or use chunk_size in tokens with a tokenizer-aware variant)
    chunk_overlap=50,      # preserves context across boundaries
    separators=["\n\n", "\n", ". ", " ", ""]
)
chunks = splitter.split_text(data)
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

embeddings = model.encode(chunks)
```

```python
knowledge_base = []

for chunk, embedding in zip(chunks, embeddings):
    knowledge_base.append({
        "text": chunk,
        "embedding": embedding
    })
```

```python
query = "explain CNN architecture"

query_embedding = model.encode(query)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

scores = cosine_similarity(
    [query_embedding],
    embeddings
)[0]
```

```python
best_index = scores.argmax()

context = chunks[best_index]
```

```python
import numpy as np

# Get the indices of the top N scores
N = 3 # Number of top chunks to retrieve
top_indices = np.argsort(scores)[-N:][::-1]

# Combine the text from these top chunks into a richer context
retrieved_contexts = [chunks[i] for i in top_indices]
context = "\n\n".join(retrieved_contexts)

print("--- New Context ---")
print(context)
print("-------------------")
```

```python
prompt = f"""
Answer using ONLY the context below.

Context:
{context}

Question:
{query}
"""
```

```python
def retrieve_top_k_chunks(query_text, k):
    query_embedding = model.encode(query_text)
    scores = cosine_similarity([query_embedding], embeddings)[0]
    top_k_indices = np.argsort(scores)[-k:][::-1]
    return top_k_indices.tolist()
```

```python
def hit_rate_at_k(test_queries, retrieve_fn, k):
  hits = 0
  all_retrieved_chunks = [] # To store all retrieved lists

  for query, correct_chunk_id in test_queries:
        retrieved = retrieve_fn(query, k)
        all_retrieved_chunks.append(retrieved)

        print(f"\nQuery: {query}")
        print(f"Expected: {correct_chunk_id}")
        print(f"Retrieved: {retrieved}")

        if correct_chunk_id in retrieved:
            print("✅ HIT")
            hits += 1
        else:
            print("❌ MISS")

  return {
      "hit_rate": hits / len(test_queries),
      "all_retrieved_chunks": all_retrieved_chunks,
  }
```

```python
result_dict = hit_rate_at_k(test_queries, retrieve_top_k_chunks, k=5)
print(f"Hit Rate at k={3}: {result_dict['hit_rate']}")
print(f"All Retrieved Chunk Lists: {result_dict['all_retrieved_chunks']}")
```

```python
def get_rank(result_dict,test_queries):
  index = []
  for j in range(len(result_dict['all_retrieved_chunks'])):
    for i in range(len(result_dict['all_retrieved_chunks'][j])):
      if result_dict['all_retrieved_chunks'][j][i] == test_queries[j][1]:
        index.append(i+1)
  return index
get_rank(result_dict,test_queries)
```

```python
reciprocal = [1/i for i in get_rank(result_dict,test_queries)]
reciprocal
```
